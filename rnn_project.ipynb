## Feature Encoding

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label = LabelEncoder()
```

```python
data['Sentiment'] = label.fit_transform(data['Sentiment'])
```

```python
for index, label1 in enumerate(label.classes_):
    print(index,label1)
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

Tokenizer need to handle out or vocab words

```python
tokenizer = Tokenizer(oov_token=None)
```

```python
tokenizer.fit_on_texts(data['Tweet'])
```

```python
tokenizer.word_index ## All words in vocab
```

```python
tokenizer.word_counts ## how many time a vocab word wa repeared, dictionary
```

```python
tokenizer.document_count ## Number of rows
```

```python
len(tokenizer.word_index) ## Unique word in vocab
```

```python
type(tokenizer.word_index)
```

```python
## Savde rows data in sequences
sequences = tokenizer.texts_to_sequences(data['Tweet'])
```

```python
sequences
```

### Padding - make size similar for all the tweets

```python
from keras.utils import pad_sequences
```

```python
sequences = pad_sequences(sequences,padding='pre')
```

```python
sequences
```

## Model Buiding without Embedding Layer

```python
from tensorflow.keras.layers import Dense, SimpleRNN, Embedding, Input
from tensorflow.keras.models import Sequential
from sklearn.model_selection import train_test_split
```

```python
y = data['Sentiment'].values
```

```python
x= sequences
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
model = Sequential()
model.add(SimpleRNN(32,input_shape=(166,1),return_sequences=False))
model.add(Dense(3,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',metrics=['accuracy'],loss='sparse_categorical_crossentropy')
```

```python
model.fit(x_train,y_train,epochs=10,validation_data=(x_test,y_test),batch_size=64)
```

## Model Buiding with Embedding Layer

```python
vocab_size =  len(tokenizer.word_index)+1
```

```python
vocab_size
```

```python
model2 = Sequential()
model2.add(Input(shape=(166,)))
model2.add(Embedding(input_dim=vocab_size,output_dim=64))
model2.add(Dense(3,activation='softmax'))
```

```python
model2.summary()
```

```python
### Add everything
```

```python
model_final = Sequential([
Embedding(input_dim=vocab_size,output_dim=128,input_shape=(166,),mask_zero=True),
SimpleRNN(32),
Dense(3,activation='softmax')
])
```

```python
model_final.summary()
```

```python
model_final.compile(optimizer='adam',metrics = ['accuracy'],loss='sparse_categorical_crossentropy')
```

```python
model_final.fit(x_train,y_train,epochs=50,validation_data=(x_test,y_test),batch_size=64)
```

```python
import pickle
with open('tokenizer.pkl','wb') as file:
    pickle.dump(tokenizer,file)
```

```python
model_final.save('model.h5')
```
